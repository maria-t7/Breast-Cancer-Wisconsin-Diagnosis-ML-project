```python
dataset = fetch_ucirepo(id=17)
X = dataset.data.features
y = dataset.data.targets
df = pd.concat([X, y], axis=1)
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Diagnosis', data=df, palette='viridis')
plt.title('Count of Benign (B) vs Malignant (M)')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['radius1'], kde=True, color='teal')
```

```python
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='Diagnosis', y='texture1', data=df, palette='Set2')
plt.title('Bivariate: Texture vs Diagnosis')
```

```python
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='radius1', y='texture1', hue='Diagnosis', data=df, alpha=0.7)
plt.title('Multivariate: Radius vs Texture')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
correlation_matrix = X.corr()
sns.heatmap(correlation_matrix, cmap='coolwarm', annot=False)
plt.title('Correlation Heatmap: Finding Redundant Features')
plt.show()
```
